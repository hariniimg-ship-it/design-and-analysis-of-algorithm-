```python
import tkinter as tk
from tkinter import messagebox

INF = float("inf")

# Real-life use: delivery route optimization
CITIES = ["Warehouse", "Hospital", "Mall", "Office", "Airport"]

DEFAULT_COST = [
    [INF, 10, 8, 9, 7],
    [10, INF, 10, 5, 6],
    [8, 10, INF, 8, 9],
    [9, 5, 8, INF, 6],
    [7, 6, 9, 6, INF]
]

def tsp_brute_force(cost):
    from itertools import permutations
    n = len(cost)
    best_cost = INF
    best_path = None

    for perm in permutations(range(1, n)):
        path = [0] + list(perm) + [0]
        total = sum(cost[path[i]][path[i + 1]] for i in range(n))
        if total < best_cost:
            best_cost = total
            best_path = path

    return best_path, best_cost

def calculate():
    try:
        values = []
        for entry in entries:
            row = []
            for e in entry:
                value = e.get().strip().upper()
                row.append(INF if value == "INF" else float(value))
            values.append(row)

        for i in range(5):
            values[i][i] = INF

        path, total = tsp_brute_force(values)

        tour = "  →  ".join(CITIES[i] for i in path)

        details = []
        for i in range(len(path) - 1):
            u, v = path[i], path[i + 1]
            details.append(
                f"{CITIES[u]} → {CITIES[v]}   :   {values[u][v]:g}"
            )

        result.config(
            text=f"OPTIMAL DELIVERY TOUR\n\n"
                 f"{tour}\n\n"
                 f"Minimum Travel Cost : {total:g}\n\n"
                 f"ROUTE DETAILS\n"
                 + "\n".join(details)
        )

        status.config(text="Status: Optimal route calculated successfully")

    except Exception:
        messagebox.showerror(
            "Invalid Input",
            "Please enter valid numbers in the cost matrix."
        )

def reset():
    for i in range(5):
        for j in range(5):
            entries[i][j].delete(0, tk.END)
            entries[i][j].insert(
                0,
                "INF" if DEFAULT_COST[i][j] == INF else str(DEFAULT_COST[i][j])
            )

    result.config(text="Enter travel costs and click Optimize.")
    status.config(text="Status: Ready")

root = tk.Tk()
root.title("SmartRoute - TSP Route Optimizer")
root.geometry("900x650")
root.configure(bg="#EEF4FF")

# Header
header = tk.Frame(root, bg="#0F4C81", height=90)
header.pack(fill="x")

tk.Label(
    header,
    text="SMARTROUTE",
    font=("Segoe UI", 24, "bold"),
    bg="#0F4C81",
    fg="white"
).pack(pady=(12, 0))

tk.Label(
    header,
    text="Travelling Salesman Problem • Route Optimization",
    font=("Segoe UI", 10),
    bg="#0F4C81",
    fg="#D9F0FF"
).pack()

tk.Label(
    root,
    text="Real-Life Application: Find the lowest-cost route for a delivery vehicle",
    font=("Segoe UI", 11, "bold"),
    bg="#EEF4FF",
    fg="#174A73"
).pack(pady=12)

# Matrix card
matrix_card = tk.Frame(root, bg="white", bd=1, relief="solid")
matrix_card.pack(padx=30, pady=5)

tk.Label(
    matrix_card,
    text="Travel Cost Matrix",
    font=("Segoe UI", 13, "bold"),
    bg="white",
    fg="#0F4C81"
).grid(row=0, column=0, columnspan=6, pady=12)

tk.Label(matrix_card, text="", bg="white", width=13).grid(row=1, column=0)

for j, city in enumerate(CITIES):
    tk.Label(
        matrix_card,
        text=city,
        font=("Segoe UI", 9, "bold"),
        bg="#DCEEFF",
        fg="#0F4C81",
        width=12,
        pady=5
    ).grid(row=1, column=j + 1, padx=1, pady=1)

entries = []

for i, city in enumerate(CITIES):
    tk.Label(
        matrix_card,
        text=city,
        font=("Segoe UI", 9, "bold"),
        bg="#DCEEFF",
        fg="#0F4C81",
        width=12,
        pady=5
    ).grid(row=i + 2, column=0, padx=1, pady=1)

    row = []
    for j in range(5):
        e = tk.Entry(
            matrix_card,
            width=12,
            justify="center",
            font=("Consolas", 10),
            bg="#F8FBFF"
        )
        e.grid(row=i + 2, column=j + 1, padx=1, pady=1)
        e.insert(
            0,
            "INF" if DEFAULT_COST[i][j] == INF else str(DEFAULT_COST[i][j])
        )
        row.append(e)

    entries.append(row)

# Buttons
buttons = tk.Frame(root, bg="#EEF4FF")
buttons.pack(pady=15)

tk.Button(
    buttons,
    text="Optimize Route",
    command=calculate,
    bg="#16A34A",
    fg="white",
    font=("Segoe UI", 11, "bold"),
    width=18,
    relief="flat",
    cursor="hand2"
).grid(row=0, column=0, padx=7)

tk.Button(
    buttons,
    text="Reset",
    command=reset,
    bg="#F59E0B",
    fg="white",
    font=("Segoe UI", 11, "bold"),
    width=15,
    relief="flat",
    cursor="hand2"
).grid(row=0, column=1, padx=7)

# Result card
result_card = tk.LabelFrame(
    root,
    text="Optimization Result",
    font=("Segoe UI", 11, "bold"),
    bg="white",
    fg="#0F4C81",
    padx=15,
    pady=10
)
result_card.pack(fill="both", expand=True, padx=30, pady=5)

result = tk.Label(
    result_card,
    text="Enter travel costs and click Optimize.",
    font=("Consolas", 11),
    bg="white",
    fg="#263238",
    justify="left",
    anchor="nw"
)
result.pack(fill="both", expand=True)

status = tk.Label(
    root,
    text="Status: Ready",
    bg="#0F4C81",
    fg="white",
    anchor="w",
    padx=15,
    pady=7,
    font=("Segoe UI", 9)
)
status.pack(fill="x", side="bottom")

root.mainloop()
```
