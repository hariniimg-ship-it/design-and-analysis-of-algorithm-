```python
import tkinter as tk
from tkinter import ttk,messagebox
import random

comparison_count=0

def dc(arr,l,h):
    global comparison_count
    if l==h:return arr[l],arr[l]
    if h==l+1:
        comparison_count+=1
        return (arr[l],arr[h]) if arr[l]<arr[h] else (arr[h],arr[l])
    m=(l+h)//2
    a,b=dc(arr,l,m); c,d=dc(arr,m+1,h)
    comparison_count+=2
    return (a if a<c else c),(b if b>d else d)

def naive(arr):
    mn=mx=arr[0]; comp=0
    for x in arr[1:]:
        comp+=1
        if x<mn: mn=x
        comp+=1
        if x>mx: mx=x
    return mn,mx,comp

def sample():
    nums=[random.randint(10,99) for _ in range(10)]
    entry.delete(0,tk.END)
    entry.insert(0,",".join(map(str,nums)))
    result.config(text="")

def clear():
    entry.delete(0,tk.END)
    result.config(text="")

def analyze():
    global comparison_count
    try:
        arr=list(map(int,entry.get().split(",")))
    except:
        messagebox.showerror("Invalid Input","Enter comma-separated integers.")
        return
    comparison_count=0
    mn,mx=dc(arr,0,len(arr)-1)
    dcc=comparison_count
    _,_,nc=naive(arr)
    result.config(text=f"Minimum Stock : {mn}\nMaximum Stock : {mx}\n\nDivide & Conquer Comparisons : {dcc}\nNaive Comparisons : {nc}")

root=tk.Tk()
root.title("Inventory Stock Analyzer")
root.geometry("720x450")
root.configure(bg="#EAF4FF")

tk.Label(root,text="INVENTORY STOCK ANALYZER",bg="#1565C0",fg="white",
font=("Segoe UI",18,"bold"),pady=12).pack(fill="x")

frm=tk.Frame(root,bg="#EAF4FF")
frm.pack(pady=20)

tk.Label(frm,text="Inventory Values",bg="#EAF4FF",font=("Segoe UI",11,"bold")).pack(anchor="w")
entry=tk.Entry(frm,width=70,font=("Consolas",11))
entry.pack(pady=5)

bf=tk.Frame(frm,bg="#EAF4FF")
bf.pack(pady=10)

tk.Button(bf,text="Generate Sample",bg="#FB8C00",fg="white",width=18,command=sample).grid(row=0,column=0,padx=5)
tk.Button(bf,text="Analyze",bg="#43A047",fg="white",width=18,command=analyze).grid(row=0,column=1,padx=5)
tk.Button(bf,text="Clear",bg="#E53935",fg="white",width=18,command=clear).grid(row=0,column=2,padx=5)

card=tk.LabelFrame(root,text="Analysis Result",font=("Segoe UI",11,"bold"),bg="white",padx=15,pady=15)
card.pack(fill="both",expand=True,padx=30,pady=15)

result=tk.Label(card,bg="white",justify="left",font=("Consolas",12))
result.pack(anchor="w")

sample()
root.mainloop()
```
