```python
import tkinter as tk
from tkinter import ttk, messagebox

def safe(board, row, col):
    for r in range(row):
        c = board[r]
        if c == col or abs(r-row) == abs(c-col):
            return False
    return True

def solve_n_queens(n):
    board = [-1] * n
    solutions = []
    backtracks = [0]

    def backtrack(row):
        if row == n:
            solutions.append(board[:])
            return
        for col in range(n):
            if safe(board, row, col):
                board[row] = col
                backtrack(row + 1)
                board[row] = -1
                backtracks[0] += 1

    backtrack(0)
    return solutions, backtracks[0]

root = tk.Tk()
root.title("N-Queens Visualizer | DAA Mini Project")
root.geometry("850x650")
root.configure(bg="#EEF2FF")

tk.Label(root, text="N-QUEENS VISUALIZER",
         font=("Segoe UI", 22, "bold"),
         bg="#5B21B6", fg="white", pady=15).pack(fill="x")

tk.Label(root, text="Backtracking Algorithm • Design & Analysis of Algorithms",
         font=("Segoe UI", 10), bg="#EEF2FF", fg="#4B5563").pack(pady=8)

top = tk.Frame(root, bg="white", bd=1, relief="solid")
top.pack(fill="x", padx=25, pady=10)

tk.Label(top, text="Board Size (N):", font=("Segoe UI", 11, "bold"),
         bg="white", fg="#312E81").pack(side="left", padx=15, pady=15)

nbox = ttk.Combobox(top, values=[4,5,6,7,8,9,10],
                    width=8, state="readonly", font=("Segoe UI", 11))
nbox.set(4)
nbox.pack(side="left", pady=15)

board = tk.Frame(root, bg="#EEF2FF")
board.pack(pady=10)

result = tk.Label(root, text="Click Solve to start.",
                  font=("Segoe UI", 11), bg="white", fg="#374151",
                  justify="left", padx=15, pady=12)
result.pack(fill="x", padx=25, pady=10)

cells = []

def draw(sol):
    global cells
    for w in board.winfo_children():
        w.destroy()
    cells = []
    n = len(sol)
    for r in range(n):
        row = []
        for c in range(n):
            bg = "#C4B5FD" if (r+c)%2 == 0 else "#7C3AED"
            x = tk.Label(board, text="", width=4, height=2,
                         font=("Segoe UI", 20, "bold"),
                         bg=bg, fg="white")
            x.grid(row=r, column=c, padx=1, pady=1)
            row.append(x)
        cells.append(row)
    for r,c in enumerate(sol):
        cells[r][c].config(text="Q", bg="#F59E0B")

def solve():
    n = int(nbox.get())
    solutions, backtracks = solve_n_queens(n)
    if not solutions:
        result.config(text=f"N={n}: No solution.")
        return
    draw(solutions[0])
    result.config(text=f"N = {n}    |    Solutions = {len(solutions)}    |    "
                       f"Backtracks = {backtracks}\nShowing Solution 1")

def show_all():
    n = int(nbox.get())
    if n > 6:
        messagebox.showinfo("Large Board",
                            "For N > 6, use Solve. There may be many solutions.")
        return
    solutions, _ = solve_n_queens(n)
    win = tk.Toplevel(root)
    win.title(f"All {n}-Queens Solutions")
    win.geometry("650x600")
    win.configure(bg="#EEF2FF")
    text = tk.Text(win, font=("Consolas", 11), width=65, height=32)
    text.pack(padx=15, pady=15)
    for i, sol in enumerate(solutions, 1):
        text.insert("end", f"Solution {i}: {sol}\n")
        for r in range(n):
            text.insert("end", "".join(" Q " if sol[r] == c else " . "
                                       for c in range(n)) + "\n")
        text.insert("end", "\n")
    text.config(state="disabled")

def clear():
    draw([-1] * int(nbox.get()))
    for r in range(len(cells)):
        for c in range(len(cells[r])):
            cells[r][c].config(text="")
    result.config(text="Click Solve to start.")

buttons = tk.Frame(root, bg="#EEF2FF")
buttons.pack(pady=8)

tk.Button(buttons, text="Solve", command=solve, width=15,
          bg="#7C3AED", fg="white", font=("Segoe UI", 11, "bold"),
          relief="flat").grid(row=0, column=0, padx=6)

tk.Button(buttons, text="View All Solutions", command=show_all, width=18,
          bg="#2563EB", fg="white", font=("Segoe UI", 11, "bold"),
          relief="flat").grid(row=0, column=1, padx=6)

tk.Button(buttons, text="Clear", command=clear, width=15,
          bg="#EF4444", fg="white", font=("Segoe UI", 11, "bold"),
          relief="flat").grid(row=0, column=2, padx=6)

tk.Label(root, text="Applications: scheduling • resource allocation • constraint solving • AI search",
         font=("Segoe UI", 9), bg="#EEF2FF", fg="#6B7280").pack(pady=10)

draw([-1,-1,-1,-1])
root.mainloop()
```
